```python
#Task 13 — Normalize a Tensor
#Create sensor data:
#X = np.random.randn(100, 20, 5)
#Calculate the mean and standard deviation for each sensor.
#Normalize the tensor using:
# x_normalized=(x-μ)/σ
#Then verify:
#mean ≈ 0
#standard deviation ≈ 1
# 1. Create sensor data (100 samples, 20 timesteps, 5 sensors)
np.random.seed(42)
X = np.random.randn(100, 20, 5)

# 2. Calculate mean (μ) and standard dev (σ) for each sensor
# Use axis=(0, 1) to reduce across samples and timesteps, leaving 5 values
# keepdims=True ensures proper shape alignment for broadcasting later
mean = X.mean(axis=(0, 1), keepdims=True)
std = X.std(axis=(0, 1), keepdims=True)

# 3. Normalize the tensor
X_normalized = (X - mean) / std

# 4. Verify results (should be ≈ 0 and ≈ 1)
verified_mean = X_normalized.mean(axis=(0, 1))
verified_std = X_normalized.std(axis=(0, 1))

print("Original shape: ", X.shape)
print("Calculated Mean per sensor:\n", mean.squeeze())
print("Calculated Std per sensor:\n", std.squeeze())
print("\n--- Verification ---")
print("Normalized Mean (Expects ~0):\n", verified_mean)
print("Normalized Std (Expects ~1):\n", verified_std)
```

```python
#Task 14 — Flattening
#Create an image:
#(28, 28, 3)
#Flatten it into:
#(2352,)
#Then create a batch of 64 images:
#(64, 28, 28, 3)
#Flatten each image while keeping the batch dimension.
#Expected shape: 	(64, 2352)
#Why shouldn't we simply flatten the entire tensor into (150528,) when working with a batch?
# --- Using PyTorch ---
# 1. Create a single image tensor (28, 28, 3)
single_img_torch = torch.randn(28, 28, 3)
# Flatten to (2352,)
flat_img_torch = torch.flatten(single_img_torch)
print("PyTorch Single Flat Shape:", flat_img_torch.shape) 

# 2. Create a batch of 64 images (64, 28, 28, 3)
import torch
import torchvision.transforms as transforms
from PIL import Image
import os

folder = "D:\images"

# Resize all images to same size
transform = transforms.Compose([
    transforms.Resize((28, 28)),
    transforms.ToTensor()
])

images = []

for file in os.listdir(folder):
    if file.endswith(".jpg") or file.endswith(".png"):
        img_path = os.path.join(folder, file)
        image = Image.open(img_path)

        tensor_img = transform(image)
        images.append(tensor_img)

batch_tensor = torch.stack(images)
# Flatten keeping batch dimension (64, 2352)
flat_batch_torch = torch.flatten(batch_torch, start_dim=1)
print("PyTorch Batch Flat Shape:", flat_batch_torch.shape)


# --- Using NumPy ---
# 1. Create a single image array
single_img_np = np.random.randn(28, 28, 3)
flat_img_np = single_img_np.flatten()
print("NumPy Single Flat Shape:", flat_img_np.shape)

# 2. Create a batch of 64 images
batch_np = np.random.randn(64, 28, 28, 3)
# Flatten keeping batch dimension
flat_batch_np = batch_np.reshape(64, -1)
print("NumPy Batch Flat Shape:", flat_batch_np.shape)
```

```python
#Task 16 — Practical Mini Challenge: Sensor Dataset
#Create a synthetic dataset representing:
#500 machines
#60 time steps
#8 sensors

#Therefore:

#X.shape = (500, 60, 8)
# Set random seed for reproducibility
np.random.seed(42)

# Create synthetic dataset: shape = (500, 60, 8)
# Axis breakdown: 0 = Machines, 1 = Time Steps, 2 = Sensors
X = np.random.rand(500, 60, 8)

# 1. Extract machine 10
# (Note: Using index 10 for explicit 0-indexed slicing; use index 9 for the literal 10th machine)
machine_10 = X[10, :, :]

# 2. Extract the first 20 time steps
first_20_timesteps = X[:, :20, :]

# 3. Extract sensor 5 from all machines
sensor_5 = X[:, :, 5]

# 4. Calculate the average value of every sensor
# Averages across all machines and time steps (axis 0 and 1)
avg_per_sensor = X.mean(axis=(0, 1))

# 5. Calculate the average value at every time step
# Averages across all machines and sensors (axis 0 and 2)
avg_per_timestep = X.mean(axis=(0, 2))

# 6. Find the machine with the highest average sensor reading
# First average time steps and sensors per machine, then find the max index
machine_averages = X.mean(axis=(1, 2))
highest_avg_machine_idx = np.argmax(machine_averages)

# 7. Normalize each sensor (Min-Max Scaling)
# Computes min and max along axes 0 and 1 while preserving the shape for broadcasting
X_min = X.min(axis=(0, 1), keepdims=True)
X_max = X.max(axis=(0, 1), keepdims=True)
X_normalized = (X - X_min) / (X_max - X_min)

# 8. Split the tensor into training and testing sets
train_X = X[:400, :, :]
test_X = X[100:, :, :]

# --- Verification Outputs ---
print(f"Dataset Shape: {X.shape}\n")
print(f"\n1. Machine 10 : {machine_10}\n")
print(f"\n2. First 20 time steps : {first_20_timesteps}\n")
print(f"\n3. Sensor 5 : {sensor_5}\n")
print(f"\n4. Average per sensor : {avg_per_sensor}\n")
print(f"\n5. Average per time step : {avg_per_timestep}\n")
print(f"\n6. Machine with highest average: Index {highest_avg_machine_idx}\n")
print(f"\n7. Normalized dataset : {X_normalized}\n")
print(f"\n8. Train : {train_X} | Test shape: {test_X}\n")
```
