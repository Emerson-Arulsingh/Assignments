```python
#Part C — Data Transformation
#Create a new column called Age_Group:
#< 18 Child
#18–40 Young Adult
#41–60 Middle Age
#> 60 Senior
df['Age_Group'] = df['Age'].apply(lambda x: 'Senior' if x > 60 else 'Middle Age' if x >= 41 and x <= 60 
                                                               else 'Young Adult' if x >= 18 and x <= 40 
                                                               else 'Child') 
print("Updated DataFrame")
print (df)
```

```python
#16. Separate Blood_Pressure into two columns:
# Systolic
# Diastolic

df["Systolic"] = df["Blood_Pressure"].str.split('/').str[0]
df["Diastolic"] = df.Blood_Pressure.str.split('/').str[1]
print(df)
```

```python
#17. Create a High_Risk column based on suitable clinical conditions.
df["HighRisk"] = df["Blood_Pressure"].str.split('/').str[0].apply(lambda x : "High Risk" if int(x) > 150 else "Normal") 
print(df)
```

```python
#18. Create a Cost_Per_Day column using treatment cost and length of stay.
df["Cost_Per_Day"] = df["Treatment_Cost"] / df["Length_of_Stay"]
print(df)
```

```python
#Part D — Analysis
#19. Calculate the average age of patients.
average_age = df["Age"].mean()
print(f"Average age of the patients ", average_age)
```

```python
#20. Find the number of patients in each department.
print("Departmentwise patients count")
print(df.groupby("Department").size())
```

```python
#21. Calculate the average treatment cost for each department.
print("average treatment cost for each department")
print(df.groupby("Department")["Treatment_Cost"].mean())
```

```python
#22. Find the average length of stay for each diagnosis.
print("average length of stay for each diagnosis.")
print(df.groupby("Diagnosis")["Length_of_Stay"].mean())
```

```python
#23. Identify the top 10 patients with the highest treatment cost.
print("top 10 patients with the highest treatment cost")
print(df.sort_values(by='Treatment_Cost', ascending=[False]).head(10))
```

```python
#24. Find the department with the highest average treatment cost.
print("department with the highest average treatment cost")
print(df.groupby("Department")["Treatment_Cost"].mean().sort_values(ascending=False).head(1))
```

```python
#25. Calculate the percentage of patients who were readmitted. 
print("percentage of patients who were readmitted.")
print(len(df[df["Readmission"] == "Yes"])/len(df)*100)
```
