```python
#Part A — Data Understanding

import pandas as pd

# 1. Load the dataset into a Pandas DataFrame.
df = pd.read_csv('ecommerce_orders_raw.csv')
#2. Display its structure and dimensions.
print(df.info())
```

```python
#3. Identify the data types of all columns.
print(df.dtypes)
```

```python
#4. Identify missing values.
print(df[df.isna().any(axis=1)])
```

```python
#5. Identify duplicate orders.
print(df[df.duplicated()])
```

```python
#6. Examine the unique values in:
#City
print(df["City"].unique())
#Product Category
print(f"\n Unique Category \n",df["Product_Category"].unique())
#Payment Method
print(f"\n Unique Payment_Method \n",df["Payment_Method"].unique())
```

```python
#Part B — Data Cleaning
#7. Handle missing ratings.
df = df.fillna({"Rating" : df["Rating"].mean()})
```

```python
#8. Standardize city names.
df["city"] = df["City"].str.strip().str.title()
```

```python
#9. Identify invalid quantities.
print(df[df["Quantity"] <= 0])
```

```python
#10. Handle duplicate orders.
df_nan = df.drop_duplicates()
print(df_nan)
```

```python
#11. Convert Order_Date into an appropriate datetime format.
df['Order_Date'] = pd.to_datetime(df['Order_Date'], dayfirst=True)
df['Order_Date'] = df['Order_Date'].dt.strftime('%d/%B/%Y')
print(df['Order_Date'])
```

```python
#Part C — Feature Engineering
#12. Create:
#Gross_Amount = Quantity × Unit_Price
df["Gross_Amount"] = df["Quantity"]*df["Unit_Price"]
print(df)
```

```python
#13.
#Create:
#Discount_Amount = Gross_Amount × Discount / 100
df["Discount_Amount"] = df["Gross_Amount"]*df["Discount"]/100
print(df)
```

```python
#14.
#Create:
#Net_Amount = Gross_Amount - Discount_Amount
df["Net_Amount"] = df["Gross_Amount"] - df["Discount_Amount"]
print(df)
```

```python
#15.
#Create a Rating_Category column:
#1–2 → Poor
#3 → Average
#4 → Good
#5 → Excellent
df['Rating_Category'] = df['Rating'].apply(lambda x: 'Excellent' if x == 5 else 'Good' if x == 4 
                                                               else 'Average' if x == 3
                                                               else 'Poor')
print(df)
```

```python
#16.
#Extract the following from Order_Date:
#Year
#Month
#Day
#Day of Week
df['Order_Date'] = pd.to_datetime(df['Order_Date'], dayfirst=True)
df['year'] = df['Order_Date'].dt.year
df['month'] = df['Order_Date'].dt.month
df['day'] = df['Order_Date'].dt.day
df['day_of_week_num'] = df['Order_Date'].dt.dayofweek   # Monday=0, Sunday=6
df['day_of_week_name'] = df['Order_Date'].dt.day_name() 
print(df)
```

```python
#Part D — Business Analysis
#17. Calculate total revenue.
print(f"Total Revenue : ", sum(df["Net_Amount"]))
```

```python
#18. Calculate total number of orders.
print(f"Total number of orders : ", len(df))
```

```python
#19. Find the average order value.
print(f"Average order value : ", df["Net_Amount"].mean())
```

```python
#20. Find total revenue by product category.
print(f"total revenue by product category ", df.groupby("Product_Category")["Net_Amount"].sum())
```

```python
#21. Find the top 10 products by revenue.
filtered_product = df.sort_values(by="Net_Amount", ascending=False).head(10)
print(filtered_product[["Product", "Unit_Price", "Net_Amount"]])
```

```python
#22. Find the city generating the highest revenue.
filtered_city_revenue = df.groupby("city")["Net_Amount"].sum()
print(f"total revenue by product category ", filtered_city_revenue.sort_values().tail(1))
```

```python
#23. Find the most frequently used payment method.
print(f"most frequently used payment method", df.groupby("Payment_Method").size().sort_values(ascending = True).tail(1))
```

```python
#24. Find the average rating for each product category.
print("Average rating for each product category.")
df.groupby("Product_Category")["Rating"].mean()
```

```python
#25.Calculate monthly revenue.
# 1. Convert the date column to standard datetime format
df['date'] = pd.to_datetime(df['Order_Date'])

# 2. Extract the year and month period (YYYY-MM)
df['month'] = df['date'].dt.to_period('M')

# 3. Group by the monthly period and sum the revenue column
monthly_revenue = df.groupby('month')['Net_Amount'].sum().reset_index()
print(monthly_revenue)
```

```python
#26. Identify the highest-revenue month.
monthly_revenue.sort_values(by="month", ascending=False).head(1)
```

```python
#27. Find the top 10 customers by total spending.
grouped_customer = df.groupby(by="Order_ID")["Net_Amount"].sum()
grouped_customer.sort_values(ascending=False).head(10)
```

```python
#28. Compare revenue generated by different cities.
df.groupby(by="city")["Net_Amount"].sum()
```

```python
#29. Find which product category has the highest average order value.
df.groupby(by="Product_Category")["Net_Amount"].max()
```
