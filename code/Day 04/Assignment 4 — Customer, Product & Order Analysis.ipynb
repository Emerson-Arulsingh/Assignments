```python
#14.Create a final consolidated DataFrame.
final_product = merged_orders.merge(products, on="Product_ID", how="inner")
print(final_product)
```

```python
#Part D — Business Calculations
#15. Create:
#Order_Value = Price × Quantity
final_product["Order_Value"] = final_product["Price"] * final_product["Quantity"]
```

```python
#16. Calculate total revenue.
print(f"Total Revenue",sum(final_product["Order_Value"]))
```

```python
#17. Calculate average order value.
print(f"average order value",final_product["Order_Value"].mean())
```

```python
#18. Find total revenue by product category.
print(f"average order value",final_product.groupby("Category")["Order_Value"].mean())
```

```python
#19. Find total revenue by city.
print(f"total revenue by city",final_product.groupby("City")["Order_Value"].sum())
```

```python
#20. Find the top 10 customers by spending.
customer_filter = final_product.groupby("Customer_Name")["Order_Value"].sum()
print(f"top 10 customers by spending", customer_filter.sort_values(ascending=False).head(10))
```

```python
#21. Find the top 10 products by revenue.
product_filter = final_product.groupby("Product_Name")["Order_Value"].sum()
print(f"top 10 products by revenue.", product_filter.sort_values(ascending=False).head(10))
```

```python
#22. Find the most frequently purchased product.
product_filter = final_product["Product_Name"].value_counts()
print(product_filter.sort_values(ascending=False).head(1))
```

```python
#23. Find the city with the highest number of customers.
city_filter = final_product["City"].value_counts()
print(city_filter.sort_values(ascending=False).head(1))
```

```python
#24. Find the city generating the highest revenue.
citywiserevenue = final_product.groupby("City")["Order_Value"].sum()
print("the city generating the highest revenue", citywiserevenue.sort_values( ascending=False).head(1))
```

```python
#Part E — Customer Analysis
#25. Divide customers into age groups.
#18–30
#31–45
#46–60
#61+
final_product["Age_Group"] = final_product["Age"].apply(lambda x : "Senior Citizen" if x >= 61
                                                        else "Adult" if x >= 46 and x <= 60
                                                        else "Senior Youth" if x>=31 and x<=45
                                                        else "Youth" if x>= 18 and x<=30
                                                        else "Child")
print(final_product)
```

```python
#26. Calculate revenue generated by each age group.
final_product.groupby("Age_Group")["Order_Value"].sum()
```

```python
#27. Compare spending between male and female customers.
final_product.groupby("Gender")["Order_Value"].sum()
```

```python
#28. Identify high-value customers.
highvaluecustomers = final_product.groupby("Customer_Name")["Order_Value"].sum()
print(f"High value customer", highvaluecustomers.sort_values(ascending = False).head(1))
```

```python
#29. Determine the percentage of customers who have placed more than one order.
customersordercount =  final_product.groupby("Customer_ID")["Order_ID"].count()
percentage = (customersordercount > 1).mean() * 100
print(percentage)
```

```python
#Part F — Advanced Analysis
#30. Create a customer spending ranking.
customersspending = final_product.groupby("Customer_Name")["Order_Value"].sum()
print(customersspending.sort_values(ascending=False))
```

```python
#31. Identify the top 10% of customers based on total spending.
print(customersspending.sort_values(ascending=False).head(int(len(customersspending)*0.1)))
```

```python
#32. Find the most profitable product category for each city.
final_product.groupby(["City","Category"])["Order_Value"].sum()
```

```python
#33. Find the most popular product in each category.
final_product.groupby(["Category","Product_Name"])["Order_Value"].sum()
```

```python
#34. Calculate monthly revenue.
# 1. Convert the date column to standard datetime format
final_product['date'] = pd.to_datetime(final_product['Order_Date'])

# 2. Extract the year and month period (YYYY-MM)
final_product['month'] = final_product['date'].dt.to_period('M')

# 3. Group by the monthly period and sum the revenue column
monthly_revenue = final_product.groupby('month')['Order_Value'].sum().reset_index()
print(monthly_revenue)
```

```python
#35. Identify the month with the highest revenue.
monthly_revenue.sort_values(by="Order_Value",ascending=False).head(1)
```
