```python
#Q1.Write a program to check whether Employee ID 1050 exists.
for employee in employees.keys():
    if employee == '1050':
        print (f"Employee Id {employee} present")
       
#Q2.Ask the user to enter an Employee ID and display whether it exists.
foundflag = False
employeeId = input("Enter the Employee Id :")
for employee in employees.keys():
    if employee == employeeId:
        print (f"Employee Id {employee} present")
        break
    else:
       foundflag = True
        
if foundflag == True:
    print (f"Employee Id {employeeId} is not present")
          
#Q3.If the Employee ID exists, display all the employee details.
foundflag = False
employeeId = input("Enter the Employee Id :")
for employee in employees.items():
    if employee[0] == employeeId:
        print (employee)
        break
    else:
       foundflag = True
        
if foundflag == True:
    print (f"Employee Id {employeeId} is not present") 
    
#Q4.If the Employee ID does not exist, display:Employee not found.
notfoundflag = False
employeeId = input("Enter the Employee Id :")
for employee in employees.keys():
    if employee == employeeId:
        foundflag = True
if notfoundflag == False:    
    print (f"Employee Id {employeeId} is not present") 
    
#Q5.Display the salary of Employee 1032.
for employee in employees.items():
    if employee[0] == '1032':
        print (f"Employee Id 1032's Salary ", employee[1]["Salary"])
        break

#Q6.Display the department of Employee 1087.
for employee in employees.items():
    if employee[0] == '1087':
        print (f"Employee Id 1087's Department ", employee[1]["Department"])
        break
       
#Q7.Print only the email address of Employee 1015.
for employee in employees.items():
    if employee[0] == '1015':
        print (f"Employee Id 1015's Email Id", employee[1]["Email"])
        break
       
#Q8.Find all employees whose email address is missing (None).
print(list([employee[1]["Name"] for employee in employees.items() if employee[1]["Email"] == None]))

#Q9.Find all employees whose phone number is missing.
print(list([employee[1]["Name"] for employee in employees.items() if employee[1]["Phone"] == None]))
 
#Q10.Count how many employees have incomplete records (missing email or phone).
print(len([1 for employee in employees.items() if employee[1]["Email"] or employee[1]["Phone"] == None]))
 
#Q11.Replace all missing email addresses with: Not Available
for employee in employees.items():
    if employee[1]["Email"]  == None or employee[1]["Email"] is None: 
        employee[1]["Email"] = "Not Available"
print(list([employee[1]["Name"] for employee in employees.items() if employee[1]["Email"] == None]))

#Q12.Find all employees working in the IT department.
print(list([employee[1]["Name"] for employee in employees.items() if employee[1]["Department"] == 'IT']))

#Q13.Find all employees working in the Finance department.
print(list([employee[1]["Name"] for employee in employees.items() if employee[1]["Department"] == 'Finance']))

#Q14.Display the names of all employees managed by Manager_4.
print(list([employee[1]["Name"] for employee in employees.items() if employee[1]["Manager"] == 'Manager_4']))

#Q15.Find all employees older than 50 years.
print(list([employee[1]["Name"] for employee in employees.items() if employee[1]["Age"] > 50]))

#Q16.Find all employees younger than 30 years.
print(list([employee[1]["Name"] for employee in employees.items() if employee[1]["Age"] < 30]))

#Q17.Find the employee with the highest salary.
maxSal = 0
name = ""
for employee in employees.items():
    if employee[1]["Salary"] > maxSal:
        maxSal = employee[1]["Salary"]
        name = employee[1]["Name"]
print (f"Employee {name} has the maximum salary with {maxSal}")        

#Q18.Find the employee with the lowest salary.
minSal = 100000000
name = ""
for employee in employees.items():
    if employee[1]["Salary"] < minSal:
        minSal = employee[1]["Salary"]
        name = employee[1]["Name"]
print (f"Employee {name} has the minimum salary with {minSal}") 

#Q19.Calculate the average salary of all employees.
totsal = 0
for employee in employees.items():
        totsal += employee[1]["Salary"]
avg = totsal/ len(employees)        
print (f"Average salary of employees is {avg}") 

#Q20.Display employees earning more than ₹100,000.
print(list([employee[1]["Name"] for employee in employees.items() if employee[1]["Salary"] > 100000]))

#Q21.Display employees earning less than ₹50,000.
print(list([employee[1]["Name"] for employee in employees.items() if employee[1]["Salary"] < 50000]))

#Q22.Count how many employees belong to each department.
from collections import Counter

dept_counts = Counter([emp[1]["Department"] for emp in employees.items()])

print("HR :",dept_counts["HR"])  
print("Finance :",dept_counts["Finance"]) 
print("IT :",dept_counts["IT"]) 
print("Sales :",dept_counts["Sales"]) 
print("Marketing :",dept_counts["Marketing"]) 
print("Operations :",dept_counts["Operations"]) 
#Example output

#HR : 18
#Finance : 15
#IT : 22
#Sales : 17
#Marketing : 14
#Operations : 14

#Q23.Count how many employees are managed by each manager.
from collections import defaultdict

managercount = defaultdict(int)

for employeeid, employeedet in employees.items():
    managercount[employeedet["Manager"]] += 1

print(dict(managercount))

#Q24. Find which department has the largest number of employees.
departmentcount = defaultdict(int)

for employeeid, employeedet in employees.items():
    departmentcount[employeedet["Department"]] += 1

deptdic = dict(departmentcount)
print(deptdic)
deptheadcount = 0
for key, value in deptdic.items():
    if value > deptheadcount:
        deptheadcount = value
        deptName = key
print(f"Highest headCount in the department {deptName}")

#Q25.Increase the salary of every employee by 10%.
for employeeid, employeedet in employees.items():
    employeedet["Salary"] *= 1.1
print(employees)   

#Q26.Update the department of Employee 1040 to Research.
employees["1040"]["Department"] = "Research"
print(employees["1040"])

#Q27.Add a new employee to the dictionary.
#employees.update({"2000":"Name": "Employee_2000", "Department": "Marketing", "Age": 40, "Salary": 80859,"Email": "employee2000@company.com", "Phone": "9891139541"})

#Q28.Delete Employee 1018 from the dictionary.
employees.pop("1018")

#Q29.Update the phone number of Employee 1075.
employees["1075"]["Phone"]  = "+44"
print(employees["1075"]["Phone"])
```

```python
#Q37. Identify employees with both email and phone number missing.
from collections import defaultdict

empdtl = defaultdict(list)
for item in employees.items():
    if item[1]['Phone'] is None:
        empdtl[item[1]["Phone"]].append(item)
    if item[1]['Email'] is None:
        empdtl[item[1]["Email"]].append(item)
print(dict(empdtl).items())
print(" ")

#Q38. Find duplicate phone numbers (if any).
from collections import defaultdict

empphone = defaultdict(list)
for item in employees.items():
    empphone[item[1]['Phone']].append(item)
for phone1 in empphone.items():
    if len(phone1[1]) > 1:
        print(phone1[1][1])
#Q39. Find employees whose email domain is company.com.
print("Employees with email id company.com")
print("-----------------------------------")
for emp in employees.items(): 
    if emp[1]["Email"] is not None:
        if "company.com" in emp[1]["Email"]:
            print(emp[1]["Name"])
        
#Q40. Create a new dictionary containing only employees from the IT department.
from collections import defaultdict

empdtl = defaultdict(list)
for item in employees.items():
    if item[1]['Phone'] is None:
        empdtl[item[1]["Phone"]].append(item)
    if item[1]['Email'] is None:
        empdtl[item[1]["Email"]].append(item)
print(dict(empdtl).items())
print(" ")
```
