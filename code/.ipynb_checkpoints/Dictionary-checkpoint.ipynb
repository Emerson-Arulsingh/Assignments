```python
students_info = {
    "Aarav": {"age": 20, "major": "Computer Science", "gpa": 3.8},
    "Bhavya": {"age": 22, "major": "Mathematics", "gpa": 3.4},
    "Chirag": {"age": 19, "major": "Physics", "gpa": 3.9},
    "Divya": {"age": 21, "major": "Biology", "gpa": 3.7},
    "Esha": {"age": 23, "major": "Chemistry", "gpa": 3.5}
}
```

```python
# 1. What is Aarav's major? Write the Python code to get it.
print(f"Aarav major is",students_info["Aarav"]["major"], "and his age ", students_info["Aarav"]["age"])
```

```python
# 2. How old is Chirag? Write the Python code to find his age.
print(f"Chirag old is",students_info["Chirag"]["age"])
```

```python
# 3. Get Bhavya's GPA using Python.
print(f"Bhavya old is",students_info["Bhavya"]["gpa"])
```

```python
# 4. Add a new student "Farhan" with age 24, major "Engineering", and GPA 3.6.
students_info.update({"Farhan":{"age": 24, "major": "Engineering", "gpa": 3.6}})
print(students_info)
```

```python
# 5. Update Divya's GPA to 3.9.
students_info["Divya"]["gpa"] = 3.9
print(f"Divya old is",students_info["Divya"]["gpa"])
```

```python
# 6. Write Python code to list all the student names.
A = (list(student for student in students_info.keys()))
print (A)
```

```python
# 7. Write Python code to find the average GPA of all students.
avggpa = ((sum(student["gpa"] for student in students_info.values()))/len(students_info))
print(f"Average GPA", avggpa)
```

```python
# 8. Remove Esha from the dictionary.
students_info.pop("Esha")
print(students_info)
```

```python
# 9. Check if "Gaurav" is in the dictionary.
if "Gaurav" in students_info:
    print("Gaurav exists")
else:
    print("Gaurav not exists")
    
```

```python
# 10. Write Python code to display the major of each student in a neat format.
for student in students_info.items():
    print(f"", student[0].ljust(10," "), student[1]["major"])
```
