```python
# Question 8  

#Create the following nested dictionary.

#students = {
#    "Student1": {
#        "Name": "John",
#        "Age": 22
#    },
#    "Student2": {
#        "Name": "Alice",
#        "Age": 21
#    },
#    "Student3": {
#        "Name": "David",
#        "Age": 20
#    }
#}

#Write a program to

 #   Print the name of Student2.
 #   Print the age of Student3.
 #   Display all student names using a loop.

students = {
    "Student1": {
        "Name": "John",
        "Age": 22
    },
    "Student2": {
        "Name": "Alice",
        "Age": 21
    },
    "Student3": {
        "Name": "David",
        "Age": 20
    }
}

print(students["Student2"]["Name"])
print(students["Student3"]["Age"])
      
for outer_key, inner_dict in students.items():
    for inner_key, value in inner_dict.items():
        if "Name" in inner_key:
            print(f"{value}")
```

```python
# Question 9  

#Create a dictionary of five products.

#Example

#    Laptop
#    Mouse
#    Keyboard
#    Monitor
#    Printer

#Each product should contain

#    Product Name
#    Price
#    Quantity

#Write a program to

#    Display all products.
#    Print the product with the highest price.
#    Calculate the total inventory value

#(Formula:  Inventory Value = Price × Quantity)

products = {"Laptop" : {"Product Name" : "Dell", "Price" : 1000, "Quantity" : 1},
           "Mouse" : {"Product Name" : "Lenovo", "Price" : 2000, "Quantity" : 1},
           "Keyboard" : {"Product Name" : "TVS Gold", "Price" : 9000, "Quantity" : 1},
           "Monitor" : {"Product Name" : "Samsung", "Price" : 4000, "Quantity" : 1},
           "Printer" : {"Product Name" : "HP", "Price" : 5000, "Quantity" : 1}}

for outer_key, inner_dict in products.items():
    for inner_key, value in inner_dict.items():
        if "Product Name" in inner_key:
            print(f"{value}")

highValue = 0;
for outer_key, inner_dict in products.items():
    for inner_key, value in inner_dict.items(): 
        if "Price" in inner_key:
            if highValue < value:
                highValue = value
print(highValue)       

print ()

total_amount = 0;
price = 0;
quantity = 0;
for outer_key, inner_dict in products.items():
    for inner_key, value in inner_dict.items():
        if "Price" in inner_key:
            price = value
        if "Quantity" in inner_key:
            quantity = value    
    total_amount = total_amount + (price * quantity)
print(total_amount)  

            
```

```python
# Question 10

#Create a dictionary of five students and their marks.

#Example

#    Alice : 85
#    Bob : 92
#    Charlie : 78
#    David : 88
#    Emma : 95

#Write a Python program to

#    Find the student with the highest marks.
#    Find the student with the lowest marks.
#    Calculate the average marks.
#    Display students who scored more than 85.
student_Marks = {"Alice" : 85, "Bob" : 92, "Charlie" : 78, "David" : 88, "Emma" : 95}

highest_score = 0
lowest_score = 100
lowest_name = ""
highest_name = ""
total_score = 0
for key, value in student_Marks.items():
    if highest_score < value:
        highest_score = value
        highest_name = key

for key, value in student_Marks.items():
    if lowest_score > value:
        lowest_score = value
        lowest_name = key
        
for key, value in student_Marks.items():
    total_score += value;

        
print ("Highest score belongs to "+ highest_name)
print ("Lowest score belongs to "+ lowest_name)
print(total_score/5)    

for key, value in student_Marks.items():
    if value > 85:
        print(key)
```
