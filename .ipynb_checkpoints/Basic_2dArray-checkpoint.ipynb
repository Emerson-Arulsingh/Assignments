```python
import numpy as np

arr = np.array([
    [10, 11, 12, 13, 14, 15],
    [20, 21, 22, 23, 24, 25],
    [30, 31, 32, 33, 34, 35],
    [40, 41, 42, 43, 44, 45],
    [50, 51, 52, 53, 54, 55],
    [60, 61, 62, 63, 64, 65]
])

print(arr.shape)
```

```python
# 2D Array Indexing & Slicing


#Extract the following using a single indexing/slicing statement
#1. Every alternate row
print("Every alternate row \n", arr[::2], :)
```

```python
# 2. Every alternate column
print("Every alternate Column \n", arr[:, ::2])
```

```python
#3. First column
print("First column \n", arr[:, 1])
```

```python
#4. Last column
print("Last column \n", arr[:, 5])
```

```python
#5. Middle row
print("Middle row \n", arr[3, :])
```

```python
#6. Middle column
print("Middle column \n", arr[:, 5])
```

```python
#7. First three rows
print("First three rows \n", arr[0:3, :])
```

```python
#8. Last three rows
print("Last three rows \n", arr[3:7, :])
```

```python
#9. First three columns
print("First three columns \n", arr[:, 0:3])
```

```python
#10. Last three columns
print("First three columns \n", arr[:, 3:6])
```

```python
#11. Reverse all rows
print("First three columns \n", arr[::-1, :])
```

```python
#12. Reverse all columns
print("First three columns \n", arr[:, ::-1])
```

```python
#13. Reverse rows and columns simultaneously
print("First three columns \n", arr[::-1, ::-1])
```

```python
#Write a single indexing/slicing statement to extract:
print(arr[1:4, 2:5])
```

```python
print(arr[:, 4:6])
```

```python
print(arr[::2, 1::2])
```
